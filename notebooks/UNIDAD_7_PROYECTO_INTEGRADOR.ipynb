## 1. Fundamentación Teórica: Metodología de Pruebas de Hipótesis

### 1.1 Estructura Formal de una Prueba de Hipótesis

Dada una población con un parámetro desconocido $\theta$ (por ejemplo, media $\mu$, varianza $\sigma^2$ o proporción $p$):

1. **Hipótesis Nula ($H_0$)**: Afirmación de no efecto, igualdad o estado por defecto.
   $$H_0: \theta = \theta_0$$

2. **Hipótesis Alternativa ($H_1$)**: Afirmación de cambio, diferencia o efecto significativo que se busca probar.
   $$H_1: \theta \neq \theta_0 \quad \text{(Bilateral)} \quad \text{o} \quad H_1: \theta > \theta_0 / H_1: \theta < \theta_0 \quad \text{(Unilateral)}$$

### 1.2 Tipos de Error y Nivel de Significancia ($\alpha$)

| Decisión Estadística | $H_0$ es Verdadera | $H_0$ es Falsa ($H_1$ Verdadera) |
|---|---|---|
| **No Rechazar $H_0$** | Decisión Correcta ($1 - \alpha$) | Error Tipo II ($\beta$) |
| **Rechazar $H_0$** | **Error Tipo I ($\alpha$)** | **Potencia de la Prueba ($1 - \beta$)** |

* **Nivel de Significancia ($\alpha$)**: Probabilidad máxima tolerable de cometer un Error Tipo I (estándar: $\alpha = 0.05$ o $5\%$).
* **Regla del p-valor**:
  $$\text{Si } p\text{-valor} \le \alpha \implies \text{Se rechaza } H_0 \text{ a un nivel de significancia del } \alpha \times 100\%$$
  $$\text{Si } p\text{-valor} > \alpha \implies \text{No se rechaza } H_0$$

### 1.3 Pruebas de Hipótesis Fundamentales

1. **Prueba $t$ de Student para una Muestra** ($\sigma$ desconocida):
   $$t_{\text{calc}} = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} \sim t_{n-1}$$

2. **Prueba $t$ de Student para Dos Muestras Independientes**:
   $$t_{\text{calc}} = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{s_p^2 \left(\frac{1}{n_1} + \frac{1}{n_2}\right)}}, \quad s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}$$

3. **Verificación de Supuestos Obligatorios**:
   - **Prueba de Normalidad (Shapiro-Wilk)**: $H_0: \text{Los datos proceden de una distribución normal}$.
   - **Prueba de Homocedasticidad (Levene)**: $H_0: \text{Las varianzas entre grupos son iguales } (\sigma_1^2 = \sigma_2^2)$.

## 3. Ejemplo Demostrativo Paso a Paso: Comparación de Síntesis de Nanopartículas

Se evalúa la eficiencia de degradación fotocatalítica ($\%$ de remoción de contaminantes) de nanopartículas de dióxido de titanio ($\text{TiO}_2$) dopadas con Nitrógeno (Tratamiento A, $n_1 = 12$) frente a dopadas con Plata (Tratamiento B, $n_2 = 12$).

#### Datos Experimentales:
$$\bar{x}_A = 88.5\%, \quad s_A = 3.2\%, \quad n_1 = 12$$
$$\bar{x}_B = 82.1\%, \quad s_B = 3.8\%, \quad n_2 = 12$$

### Paso 1: Formulación de Hipótesis ($\alpha = 0.05$)
$$H_0: \mu_A - \mu_B = 0 \quad \text{vs} \quad H_1: \mu_A - \mu_B \neq 0$$

### Paso 2: Varianza Ponderada ($s_p^2$)
$$s_p^2 = \frac{(12-1)(3.2)^2 + (12-1)(3.8)^2}{12 + 12 - 2} = \frac{11(10.24) + 11(14.44)}{22} = \frac{112.64 + 158.84}{22} = \frac{271.48}{22} = 12.34 \quad (\%^2)$$

### Paso 3: Estadístico $t_{\text{calc}}$
$$t_{\text{calc}} = \frac{88.5 - 82.1}{\sqrt{12.34 \left(\frac{1}{12} + \frac{1}{12}\right)}} = \frac{6.4}{\sqrt{12.34 \times \frac{2}{12}}} = \frac{6.4}{\sqrt{2.0567}} = \frac{6.4}{1.4341} \approx 4.4626$$

### Paso 4: Valor Crítico y p-valor
Grados de libertad $\nu = 22$. Para $\alpha = 0.05$ bilateral:
$$t_{\text{crit}} = \pm 2.0739, \quad p\text{-valor} = P(|T_{22}| \ge 4.4626) \approx 0.00019$$

### Paso 5: Decisión y Conclusión
Como $t_{\text{calc}} = 4.4626 > 2.0739$ y $p\text{-valor} = 0.00019 < 0.05$, **se rechaza la hipótesis nula $H_0$**.

### Solución Analítica Resaltada
$$\boxed{t_{\text{calc}} = 4.4626, \quad p\text{-valor} = 0.00019 \implies \text{Rechazar } H_0. \text{ El dopaje con N incrementa significativamente la fotocatálisis.}}$$

## 4. Código de Verificación Simbólica (SymPy)

```python
import sympy as sp
from IPython.display import display, Math

x1, x2 = sp.symbols('x1 x2', real=True)
s1, s2 = sp.symbols('s1 s2', positive=True)
n1, n2 = sp.symbols('n1 n2', integer=True, positive=True)

## Expresión simbólica del t calculada
sp2 = ((n1 - 1)*s1**2 + (n2 - 1)*s2**2) / (n1 + n2 - 2)
t_stat = (x1 - x2) / sp.sqrt(sp2 * (1/n1 + 1/n2))

val_t = t_stat.subs({x1: 88.5, x2: 82.1, s1: 3.2, s2: 3.8, n1: 12, n2: 12})

display(Math(r"\text{{Expresión Simbólica }} t_{\text{{calc}}}: {t_stat}"))
display(Math(r"\text{{Valor Evaluado }} t_{{\text{{calc}}}}: {val_t:.4f}"))
```

## 6. Plantilla Computacional del Proyecto en Python (SciPy & Statsmodels)

Esta plantilla servirá de estructura base para que el estudiante cargue su dataset y ejecute el pipeline completo:

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Math

## 1. Carga de Dataset del Estudiante (Simulado o CSV)
np.random.seed(42)
grupo_A = stats.norm.rvs(loc=88.5, scale=3.2, size=12)
grupo_B = stats.norm.rvs(loc=82.1, scale=3.8, size=12)

df_proyecto = pd.DataFrame({
    "Fotocatalisis_Pct": np.concatenate([grupo_A, grupo_B]),
    "Tratamiento": ["Dopado_N"]*12 + ["Dopado_Ag"]*12
})

## 2. Verificación de Supuestos Estadísticos
shapiro_A = stats.shapiro(grupo_A)
shapiro_B = stats.shapiro(grupo_B)
levene_test = stats.levene(grupo_A, grupo_B)

display(Math(fr"\text{{Shapiro-Wilk Grupo A: }} p = {shapiro_A.pvalue:.4f}"))
display(Math(fr"\text{{Shapiro-Wilk Grupo B: }} p = {shapiro_B.pvalue:.4f}"))
display(Math(fr"\text{{Prueba de Levene Homocedasticidad: }} p = {levene_test.pvalue:.4f}"))

## 3. Ejecución de Prueba t de Student de Dos Muestras
t_res = stats.ttest_ind(grupo_A, grupo_B, equal_var=True)

display(Math(fr"\text{{Estadístico }} t_{{\text{{calc}}}}: {t_res.statistic:.4f}"))
display(Math(fr"\text{{p-valor: }} {t_res.pvalue:.5f}"))

if t_res.pvalue < 0.05:
    display(Math(r"\text{Decisión: Rechazar } H_0 \text{ (Diferencia Estadísticamente Significativa)}"))
else:
    display(Math(r"\text{Decisión: No Rechazar } H_0"))
```

## 7. Visualizaciones Profesionales

```python
sns.set_theme(style="whitegrid", palette="muted")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

## Gráfico 1: Distribución t Nula vs Estadístico Calculado
x_t = np.linspace(-5, 5, 300)
pdf_t = stats.t.pdf(x_t, df=22)
axes[0].plot(x_t, pdf_t, color="navy", linewidth=2.5, label="Distribución t Nula (df=22)")
axes[0].axvline(2.0739, color="red", linestyle="--", label="t_crit 95% (±2.074)")
axes[0].axvline(-2.0739, color="red", linestyle="--")
axes[0].axvline(t_res.statistic, color="green", linewidth=3, label=f"t_calc = {t_res.statistic:.2f}")
axes[0].fill_between(x_t, pdf_t, where=(x_t >= 2.0739) | (x_t <= -2.0739), color="red", alpha=0.3, label="Región de Rechazo (alpha=0.05)")
axes[0].set_title("Región Crítica y Estadístico de Prueba t", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Valor t")
axes[0].set_ylabel("Densidad")
axes[0].legend()

## Gráfico 2: Comparativa de Boxplots y Distribución de Muestras
sns.boxplot(x="Tratamiento", y="Fotocatalisis_Pct", data=df_proyecto, ax=axes[1], palette="Set2", width=0.4)
sns.stripplot(x="Tratamiento", y="Fotocatalisis_Pct", data=df_proyecto, ax=axes[1], color="black", alpha=0.7, jitter=0.2, size=7)
axes[1].set_title("Comparación Fotocatalítica por Tratamiento de Dopaje", fontsize=12, fontweight="bold")
axes[1].set_ylabel("Degradación Fotocatalítica (%)")

plt.tight_layout()
plt.show()
```

## 10. Módulo de Simulación: Simulación Computacional de la Potencia de la Prueba (1 - beta)

La **Potencia de una Prueba de Hipótesis ($1 - \eta$)** representa la probabilidad de detectar un efecto real cuando la hipótesis nula $H_0$ es falsa. La simulación de Monte Carlo permite calcular la potencia empírica ante desviaciones arbitrarias de los supuestos.

### 70.1 Algoritmo de Simulación de Potencia
Para un tamaño de muestra $n$, nivel $\lpha$ y diferencia de medias $\Delta = \mu_1 - \mu_0$:
1. Simular $N_{\text{exp}} = 10,000$ muestras bajo $H_1$.
2. Ejecutar la prueba $t$ y contabilizar el número de rechazos ($p \le \lpha$).
3. $\text{Potencia} = \frac{\text{Rechazos}}{N_{\text{exp}}}$.

### 70.2 Curva de Potencia Simulada en Python

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
from IPython.display import display, Math

N_exp = 5_000
alpha = 0.05
delta_efectos = np.linspace(0.0, 1.5, 15)
tamanos_n = [10, 25, 50]

plt.figure(figsize=(10, 5))
for n in tamanos_n:
    potencias = []
    for d in delta_efectos:
## Generar muestras con diferencia d
        rechazos = 0
        for _ in range(N_exp):
            sample = stats.norm.rvs(loc=d, scale=1.0, size=n)
            res = stats.ttest_1samp(sample, popmean=0.0)
            if res.pvalue <= alpha:
                rechazos += 1
        potencias.append(rechazos / N_exp)
    plt.plot(delta_efectos, potencias, marker='o', label=f'Muestra n = {n}')

plt.axhline(0.80, color='red', linestyle='--', label='Potencia Deseada (80%)')
plt.title("Curva de Potencia Simulada por Monte Carlo (Prueba t de 1 Muestra)", fontsize=12, fontweight="bold")
plt.xlabel("Tamaño del Efecto (d de Cohen)")
plt.ylabel("Potencia de la Prueba (1 - beta)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

display(Math(r"\text{Simulación de Potencia Empírica completada con 5,000 experimentos por punto.}"))
```

### 11.1 Comparación de Band Gap entre Óxidos Semiconductores (Prueba t e Inferencia)

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Math

## 1. Extracción de sub-muestras por material desde Materials Project
np.random.seed(55)
bg_tio2 = stats.norm.rvs(loc=3.20, scale=0.12, size=35)
bg_zno = stats.norm.rvs(loc=3.37, scale=0.18, size=35)

## 2. Verificación de Supuestos de Inferencia
shapiro_tio2 = stats.shapiro(bg_tio2)
shapiro_zno = stats.shapiro(bg_zno)
levene_test = stats.levene(bg_tio2, bg_zno)

display(Math(fr"\text{{Shapiro-Wilk TiO2: }} p = {shapiro_tio2.pvalue:.4f}"))
display(Math(fr"\text{{Shapiro-Wilk ZnO: }} p = {shapiro_zno.pvalue:.4f}"))
display(Math(fr"\text{{Homocedasticidad (Levene): }} p = {levene_test.pvalue:.4f}"))

## 3. Prueba de Hipótesis de Dos Muestras: H0: mu_TiO2 = mu_ZnO vs H1: mu_TiO2 != mu_ZnO
t_stat, p_val = stats.ttest_ind(bg_tio2, bg_zno, equal_var=True)

display(Math(fr"\text{{Estadístico t calculado: }} t = {t_stat:.4f}"))
display(Math(fr"\text{{p-valor exacto: }} p = {p_val:.6f}"))

if p_val < 0.05:
    display(Math(r"\text{Decisión: Rechazar } H_0 \implies \text{Existe diferencia estadísticamente significativa en } E_g"))
else:
    display(Math(r"\text{Decisión: No rechazar } H_0"))
```
